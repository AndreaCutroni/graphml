Utility functions to reset the face dictionaries and transfer dictionaries by jey

```python
#This function cleans up face data by removing unnecessary information while keeping the "face_id"
def reset_dictionaries(shell):
    faces = Topology.Faces(shell)
    for i, f in enumerate(faces): # Loop through each face (i is the counter, f is the actual face)
        d = Topology.Dictionary(f)
        keys = Dictionary.Keys(d) # Get all the label names stored in this dictionary
        for key in keys:
            if not key == "face_id": # Check: is this label NOT called "face_id"? (we keep only face_id)
                d = Dictionary.RemoveKey(d, key) # If true, delete this label from the dictionary
        f = Topology.SetDictionary(f, d) # Save the cleaned dictionary back onto the face


#This function copies data between objects by matching them on a specific key value.
def transfer_dict_by_key(topologies, selectors, key):
    
    dicts = {} # create an empty storage container. {} = Dictionary (stores key-value pairs) — allows you to look things up by name
    for t in topologies:
        d = Topology.Dictionary(t)
        value = Dictionary.ValueAtKey(d, key, None) # Look up the value of the key (return None if it doesn't exist)
        if value: # If the key exists (and is not empty/None), then:
            dicts[str(value)] = t #Store this object using its key value as the lookup name Example: dicts["5"] = object

    for s in selectors:
        d = Topology.Dictionary(s)
        value = Dictionary.ValueAtKey(d, key, None)
        if value:
            f = dicts[str(value)]
            f = Topology.SetDictionary(f, d)
```

Create a grid overlay

```python
b_r = Wire.BoundingRectangle(floorplan)
d = Topology.Dictionary(b_r)
xmin = Dictionary.ValueAtKey(d, "xmin")
xmax = Dictionary.ValueAtKey(d, "xmax")
ymin = Dictionary.ValueAtKey(d, "ymin")
ymax = Dictionary.ValueAtKey(d, "ymax")
width = Dictionary.ValueAtKey(d, "width")
length = Dictionary.ValueAtKey(d, "length")
uRange = list(range(0, int(width)+6,6))
vRange = list(range(0, int(length)+6,6))

grid = Grid.EdgesByDistance(floorplan, clip=True, uRange=uRange, vRange=vRange)
```
